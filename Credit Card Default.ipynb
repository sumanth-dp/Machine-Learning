```python
x=df.iloc[:,:-1]
x
```

```python
y = df.iloc[:,-1]
y
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size = 0.2, random_state = 42, stratify=y)
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
y_train.shape
```

```python
y_test.shape
```

```python
numerical_cols = [
    'LIMIT_BAL',
    'AGE',
    'BILL_AMT1',
    'BILL_AMT2',
    'BILL_AMT3',
    'BILL_AMT4',
    'BILL_AMT5',
    'BILL_AMT6',
    'PAY_AMT1',
    'PAY_AMT2',
    'PAY_AMT3',
    'PAY_AMT4',
    'PAY_AMT5',
    'PAY_AMT6'
]

categorical_cols = [
    'SEX',
    'EDUCATION',
    'MARRIAGE',
    'PAY_0',
    'PAY_2',
    'PAY_3',
    'PAY_4',
    'PAY_5',
    'PAY_6'
]
```

```python
from sklearn.preprocessing import StandardScaler, OneHotEncoder
```

```python
scaler = StandardScaler()

x_train[numerical_cols] = scaler.fit_transform(
    x_train[numerical_cols]
)

x_test[numerical_cols] = scaler.transform(
    x_test[numerical_cols]
)
```

```python
x_train
```

```python
x_test
```

```python
encoder = OneHotEncoder(
    handle_unknown='ignore',
    sparse_output=False
)

x_train_encoded = encoder.fit_transform(
    x_train[categorical_cols]
)

x_test_encoded = encoder.transform(
    x_test[categorical_cols]
)
```

```python
x_train_encoded
```

```python
import numpy as np

x_train_num = x_train[numerical_cols].values
x_test_num = x_test[numerical_cols].values

x_train_final = np.hstack([
    x_train_num,
    x_train_encoded
])

x_test_final = np.hstack([
    x_test_num,
    x_test_encoded
])
```

```python
x_train_final
```

```python
x_test_final
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression(max_iter=1000)
#model = DecisionTreeClassifier(criterion = 'entropy', max_depth = 10, random_state=42)
#model = RandomForestClassifier(n_estimators = 20, max_depth = 10, random_state = 42)
#model = XGBClassifier(n_estimators=100, max_depth = 40, random_state = 42)
#model = LogisticRegression(max_iter=1000, class_weight='balanced')
```

```python
model.fit(x_train_final, y_train)
```

```python
y_pred = model.predict(x_test_final)
```

```python
y_pred
```

```python
print(classification_report(y_test, y_pred))
```

```python
print(confusion_matrix(y_test, y_pred))
```

```python
from sklearn.model_selection import cross_val_score

scores = cross_val_score(
    model,
    x_train_final,
    y_train,
    cv=5,
    scoring='accuracy'
)

print("CV Scores:", scores)
print("Mean Accuracy:", scores.mean())
print("Standard Deviation:", scores.std())
```
